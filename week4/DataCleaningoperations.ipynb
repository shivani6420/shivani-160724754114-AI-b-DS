```python
import pandas as pd

link="https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
data=pd.read_csv(link)

print("dataset info")
data.info()

print("\nmissing values before cleaning")
print(data.isnull().sum())

print("\nfirst few rows")
print(data.head())

age_med=data["Age"].median()
data["Age"]=data["Age"].fillna(age_med)

emb_val=data["Embarked"].mode()[0]
data["Embarked"]=data["Embarked"].fillna(emb_val)

print("\nmissing values after filling")
print(data.isnull().sum())

data=data.drop("Cabin",axis=1)

print("\ncolumns now")
print(data.columns)

rows_before=len(data)
data=data.drop_duplicates()
rows_after=len(data)

print("\nduplicates removed:",rows_before-rows_after)

data["Survived"]=data["Survived"].astype("category")
data["Pclass"]=data["Pclass"].astype("category")

print("\ndata types now")
print(data.dtypes)

data.columns=data.columns.str.lower()

print("\ncolumn names changed")
print(data.columns)

q1=data["age"].quantile(0.25)
q3=data["age"].quantile(0.75)
iqr=q3-q1

low=q1-1.5*iqr
high=q3+1.5*iqr

before=len(data)
data=data[(data["age"]>=low) & (data["age"]<=high)]
after=len(data)

print("\noutliers removed:",before-after)

data.to_csv("cleaned_titanic.csv",index=False)

print("\ncleaned file saved")

print("\npreview of cleaned data")
print(data.head())
```
