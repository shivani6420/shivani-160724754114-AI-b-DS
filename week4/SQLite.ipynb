```python
import sqlite3

db=sqlite3.connect("example.db")
cur=db.cursor()

cur.execute("""CREATE TABLE IF NOT EXISTS users(
id INTEGER PRIMARY KEY AUTOINCREMENT,
name TEXT NOT NULL,
age INTEGER,
email TEXT UNIQUE
)""")

print("table created")

cur.execute("INSERT INTO users(name,age,email) VALUES(?,?,?)",("Alice",25,"alice@example.com"))
cur.execute("INSERT INTO users(name,age,email) VALUES(?,?,?)",("Bob",30,"bob@example.com"))

db.commit()
print("data inserted")

cur.execute("SELECT * FROM users")
data=cur.fetchall()

print("\nusers list")
for i in data:
    print(i)

cur.execute("UPDATE users SET age=? WHERE name=?", (28,"Alice"))
db.commit()
print("\nupdated")

cur.execute("DELETE FROM users WHERE name=?",("Bob",))
db.commit()
print("deleted")

cur.execute("SELECT * FROM users")
data=cur.fetchall()

print("\nfinal data")
for i in data:
    print(i)

db.close()
print("\nconnection closed")
```
